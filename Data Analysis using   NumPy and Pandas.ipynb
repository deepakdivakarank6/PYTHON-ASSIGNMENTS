# Numpy Array Operations

```python
import numpy as np
import pandas as pd
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
print( temperatures_w1.shape)
print( temperatures_w1.dtype)
print( temperatures_w1.size)
```

```python
fahrenheit = (temperatures_w1 * 9/5) + 32
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)
```

```python
print("Fahrenheit:", fahrenheit)
print("Maximum Temperature:", max_temp)
print("Minimum Temperature:", min_temp)
print("Mean Temperature:", round(mean_temp, 2))
```

# Pandas Series

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
marks
```

```python
mark_1st_rank = marks.iloc[0]
top_3_ranks = marks.loc['Rank1':'Rank3']
mark_3rd_rank = marks.iloc[2]
above_90 = marks[marks > 90]

print(mark_1st_rank)
print(top_3_ranks)
print( mark_3rd_rank)
print( above_90)
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

marks['Rank1'] = 100
marks = marks.drop('Rank5')
cgpa = marks / 10
print(marks)
print(cgpa)
```

# Pandas DataFrame

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 
                        'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print(transactions.columns)
```

```python
print("\nSelected Columns ('ProductCategory' and 'Amount'):\n", transactions[['ProductCategory', 'Amount']])
```

```python
filtered_north = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print( filtered_north)
```

```python
mean_amount_by_region = transactions.groupby('Region')['Amount'].mean()
print(mean_amount_by_region)
```
